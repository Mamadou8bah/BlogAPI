## Python Lists

A **list** is an **ordered**, **mutable** (changeable) collection that can hold items of any type, including mixed types and duplicates.

### Key Points
- **Indexed and sliceable**: `lst[0]`, `lst[1:3]`
- **Allows duplicates**
- **Dynamic size**: you can `append()` or `remove()` items
- **Iterable**: can be looped over with `for`  


````python
# Empty list
empty_list = []

# List of integers
numbers = [1, 2, 3, 4, 5]

# List of strings
fruits = ['apple', 'banana', 'cherry']

# Mixed data types
mixed_list = [1, 'hello', 3.14, True]

# List of lists (nested list)
matrix = [[1, 2], [3, 4]]
````

#### Accessing List Elements
List elements are indexed starting from 0. Negative indices start from -1 (last element).


````python
fruits = ['apple', 'banana', 'cherry']

# Accessing elements by index
first_fruit = fruits[0]   # 'apple'
second_fruit = fruits[1]  # 'banana'

# Negative indexing
last_fruit = fruits[-1]   # 'cherry'
second_last = fruits[-2]  # 'banana'

# Accessing elements in a nested list
matrix = [[1, 2],
          [3, 4]]
element = matrix[0][1]    # 2
````

### Slicing Lists
- SLicing allows you to get a subset of a list



````python
numbers = [1, 2, 3, 4, 5, 6]

# Syntax: list[start:stop:step]
subset = numbers[1:4]      # [2, 3, 4]
first_three = numbers[:3]  # [1, 2, 3]
from_four = numbers[3:]    # [4, 5, 6]
every_other = numbers[::2] # [1, 3, 5]
reversed_list = numbers[::-1]  # [6, 5, 4, 3, 2, 1]
````

#### Modifying Lists
Lists are mutable, meaning their contents can be changed after creation.
#### Changing Elements

````python
numbers = [10, 20, 30]
numbers[1] = 25  # [10, 25, 30]
````

#### Adding Elements
  - append(item): Adds an item to the end of the list

````python
numbers = [1, 2, 3]
numbers.append(4)  # [1, 2, 3, 4]

# insert(index, item): Inserts an item at a specified index.
numbers.insert(1, 1.5)  # [1, 1.5, 2, 3, 4]

#remove(item): Removes the first occurrence of an item.
numbers.remove(1.5)  # [1, 2, 3, 4, 5, 6, 7, 8]
# pop(index): Removes and returns the item at the given index. If index is omitted, removes the last item.
last_item = numbers.pop()    # 8
specific_item = numbers.pop(0)  # 1
# clear(): Removes all items from the list.
numbers.clear()  # []




````

## List Operations
Concatenation (+): Combines two lists.


````python
list1 = [1, 2]
list2 = [3, 4]
combined = list1 + list2  # [1, 2, 3, 4]
combined
````

````python
# Repetition (*): Repeats a list multiple times.

zeros = [0] * 5  # [0, 0, 0, 0, 0]
zeros
````

### Iterating Over Lists
You can iterate over lists using loops.



````python
fruits = ['apple', 'banana', 'cherry']

# Using a for loop
for fruit in fruits:
    print(fruit)

# Using indices
for i in range(len(fruits)):
    print(fruits[i])

# Using enumerate to get index and value
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

###  Sorting Lists
`sort()`: Sorts the list in place.




````python
numbers = [5, 2, 9, 1]
numbers.sort()  # [1, 2, 5, 9]
numbers.sort(reverse=True)  # [9, 5, 2, 1]
# Custom sort
words = ['apple', 'banana', 'cherry']
words.sort(key=len)  # ['apple', 'banana', 'cherry'] (sorted by length)

#sorted(iterable): Returns a new sorted list without modifying the original.
numbers = [3, 1, 4, 1, 5]
sorted_numbers = sorted(numbers)  # [1, 1, 3, 4, 5]


````

### Nested Lists (2D Lists)



````python
# 2D list (matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Accessing elements
element = matrix[1][2]  # 6

# Iterating over a 2D list
for row in matrix:
    for item in row:
        print(item, end=' ')
    print()
````

## 14. Python Dictionaries

- In Python, a dictionary is a built-in data type that represents an unordered, mutable collection of key-value pairs.
-  Dictionaries are also known as associative arrays or hash maps in other programming languages.
-  They allow you to store and retrieve data efficiently based on unique keys.

### Creating Dictionaries
- Dictionaries can be created using curly braces `{}` with key-value pairs separated by colons `:`.



````python
# Empty dictionary
empty_dict = {}

# Dictionary with string keys
person = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York'
}

# Dictionary with mixed key types
employee = {
    101: 'John Doe',
    102: 'Jane Smith',
    103: 'Emily Davis'
}

# Using the dict() constructor
person = dict(name='Bob', age=25, city='Los Angeles')

# Creating a dictionary from a list of tuples
items = [('name', 'Charlie'), ('age', 28)]
person = dict(items)
````

### Accessing Dictionary Elements
You can access dictionary values by referencing their keys using square brackets `[]` or the `get()` method.

````python
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Using square brackets
name = person['name']       # 'Alice'

# Using get() method
age = person.get('age')     # 30


````

### Modifying Dictionaries
Dictionaries are `mutable`, allowing you to add, update, or remove key-value pairs.
### Adding Elements



````python
person = {'name': 'Alice', 'age': 30}

# Adding a new key-value pair
person['city'] = 'New York'
# person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Updating an existing key
person['age'] = 31
# person = {'name': 'Alice', 'age': 31, 'city': 'New York'}

# Removing elements

age = person.pop('age')  # 31
# person = {'name': 'Alice', 'city': 'New York'}

# del statement: Deletes a key-value pair
del person['name']
# person = {'city': 'New York'}

# clear(): Removes all items from the dictionary
person.clear()
# person = {}
````

### Dictionary Methods
`keys()`: Returns a view object of all keys.



````python
person = {'name': 'Alice', 'age': 30}

keys = person.keys()  # dict_keys(['name', 'age', 'city'])
keys
````

`values()`: Returns a view object of all values.



````python
values = person.values()  # dict_values(['Alice', 30, 'New York'])
````

`items()`: Returns a view object of key-value pairs as tuples.


````python
items = person.items()  # dict_items([('name', 'Alice'), ('age', 30), ('city', 'New York')])
````

### Iterating Over Dictionaries
You can iterate over keys, values, or key-value pairs.



````python
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Iterating over keys
for key in person:
    print(key)

# Iterating over values
for value in person.values():
    print(value)

# Iterating over key-value pairs
for key, value in person.items():
    print(f"{key}: {value}")
````

## 16. Python While Loops

Loops allow you to execute code multiple times.
The basic syntax of a while loop in Python is:



````python
while condition:
    # Code block to execute
````

-condition: A Boolean expression that is evaluated before each iteration. If the condition is True, the loop continues; if it is False, the loop exits.
- Code Block: The indented block of code that will execute repeatedly as long as the condition is True.

````python
i = 1
while i < 6:
    print(i)
    i += 1
````

## break and continue Statements
### break Statement
The break statement terminates the loop entirely and transfers execution to the statement immediately following the loop

````python
count = 0
while True:
    print("Count is:", count)
    count += 1
    if count >= 5:
        break  # Exit the loop
````

## continue Statement
The continue statement skips the rest of the code inside the loop for the current iteration and moves to the next iteration.

````python
count = 0
while count < 5:
    count += 1
    if count == 3:
        continue  # Skip the rest of the loop when count is 3
    print("Count is:", count)
````

### Using else with while Loops
- An optional else clause can be added to a while loop. The else block executes when the loop condition becomes False naturally, but not when the loop is terminated by a break statement.

````python
count = 0
while count < 5:
    print("Count is:", count)
    count += 1
else:
    print("Loop has ended.")
````

## Defining Functions

### Function with Parameters

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
````

### Function Parameters and Arguments
#### Positional Arguments
Arguments passed to a function based on their position.

````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8
````

### Keyword Arguments
Arguments passed to a function using parameter names as keywords.



````python
def subtract(a, b):
    return a - b

result = subtract(b=5, a=10)
print(result)  # Output: 5
````

### Default Arguments
 - Parameters with default values, which are used if no argument is provided.




````python
def power(base, exponent=2):
    return base ** exponent

print(power(5))       # Output: 25 (uses default exponent 2)
print(power(5, 3))    # Output: 125
````

*args **(Non-Keyword Variable Arguments)**
- Allows a function to accept any number of positional arguments as a tuple.



````python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

print(multiply(2, 3))          # Output: 6
print(multiply(2, 3, 4))       # Output: 24
````

### Scope of Variables
- Local Scope: Variables defined within a function are local to that function.
- Global Scope: Variables defined outside any function are global.


````python
x = "global"

def my_function():
    x = "local"
    print("Inside function:", x)

my_function()
print("Outside function:", x)
````

### Using global Keyword
-If you need to modify a global variable inside a function, use the global keyword.



````python
count = 0

def increment():
    global count
    count += 1

increment()
print(count)  # Output: 1
````

##  Python Classes/Objects

Key Concepts:
- **Class**: A blueprint or template for creating objects.
- **Object**: An instance of a class.

### Classes and Objects in Python
- In Python, everything is an object, and classes define the structure and behavior of these objects.

- **Class**: Defines a new data type.
- **Object**: An instance of a class with actual values.

### Defining a class
- You can define a class using the `class` keyword.


````python
class Robot:
    pass  # An empty class
````

Example:

````python
class Robot:
    def __init__(self, name):
        self.name = name
````

- `__init__` is a special method called a constructor.
- `self` refers to the instance of the class.

````python
robot1 = Robot("R2-D2")
robot2 = Robot("C-3PO")
````

### Instance Attributes and Methods

Attributes that are specific to each object.

````python
class Robot:
    def __init__(self, name):
        self.name = name  # Instance attribute
````

### Instance Methods
- Functions defined within a class that operate on instances of that class.



````python
class Robot:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"Hello, my name is {self.name}."
````

````python
robot = Robot("HAL 9000")
print(robot.introduce())  # Output: Hello, my name is HAL 9000.
````

````python
class Robot:
    population = 0  # Class attribute

    def __init__(self, name):
        self.name = name
        Robot.population += 1
````

````python
print(Robot.population)  # Accessing via class
print(robot.population)  # Accessing via instance
````

### Encapsulation and Access Modifiers
#### Encapsulation
- The bundling of data (attributes) and methods that operate on that data within one unit, e.g., a class.

  #### Access Modifiers in Python
- `Public`: Accessible from anywhere (`self.name`)
- `Protected`: Conventionally indicated with a single underscore (`self._name`)
- `Private`: Conventionally indicated with a double underscore (`self.__name`)
- Note: Python does not enforce access restrictions but relies on naming conventions.

````python
class Data:
    def __init__(self):
        self.public = "I am public"
        self._protected = "I am protected"
        self.__private = "I am private"

    def get_private(self):
        return self.__private
````

##  Python Modules

- A module is a file containing Python definitions and statements intended for use in other Python programs.
-Modules help organize code into manageable sections and promote code reusability.
### Why Use Modules?
- Organization: Break down complex programs into smaller, manageable, and organized files.
- Reusability: Reuse functions, classes, and variables across multiple programs.

### Creating a Module
- Any Python file (`.py` extension) can be considered a module.
- Example: Create a file named `math_utils.py`.


````python
# math_utils.py

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
````

### Using a Module
- Import the module into another script using the import statement.


````python
# main.py

import math_utils

result = math_utils.add(5, 3)
print(result)  # Output: 8
````

### NumPy Arrays

- The core of NumPy is the `ndarray` (n-dimensional array) object.
 - An ndarray is a homogeneous collection of elements indexed by non-negative integers.


````python
import numpy as np

# One-dimensional array
a = np.array([1, 2, 3])
print(a)  # Output: [1 2 3]

# Two-dimensional array
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)
# Output:
# [[1 2 3]
#  [4 5 6]]
````

##  Array Attributes

### Basic Array Operations
- Operations are applied element-wise.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)       # Output: [5 7 9]
print(a - b)       # Output: [-3 -3 -3]
print(a * b)       # Output: [4 10 18]
print(a / b)       # Output: [0.25 0.4 0.5]
print(a ** 2)      # Output: [1 4 9]
````

## Mathematical Functions
### Aggregation Functions
- Operate over an entire array or along an axis.

````python
a = np.array([[1, 2],
              [3, 4]])

print(a.sum())          # Output: 10
print(a.sum(axis=0))    # Output: [4 6]
print(a.sum(axis=1))    # Output: [3 7]
````

### Statistical Functions



````python
print(a.mean())         # Output: 2.5
print(a.std())          # Standard deviation
print(a.min())          # Minimum value
print(a.max())          # Maximum value
print(a.argmin())       # Index of minimum value
print(a.argmax())       # Index of maximum value
````

### Grouby
- The groupby command in pandas is very useful for aggregating data and performing operations on groups.
- It makes the task of splitting the Dataframe over some criteria really easy and efficient.

````python
import pandas as pd
nba = pd.read_csv("nba.csv")
nba.head()
````

- fill missing values with zero

````python
#Calculate the growth rate between 1952 and 2007 for each country:

cleannba = nba.fillna(value=0)
cleannba.head(10)
````

````python
## Average age in each team
average = cleannba.groupby("Team")["Age"].mean().round().astype(int)
average
````

````python
# number of players in each team
player_counts = cleannba.groupby("Team")["Name"].count()
player_counts
````

````python
#Total salary per team
team_salary = cleannba.groupby("Team")["Salary"].sum()
team_salary
````

````python
team_salary = cleannba.groupby("Team")["Salary"].sum().sort_values(ascending=False)
team_salary
````
